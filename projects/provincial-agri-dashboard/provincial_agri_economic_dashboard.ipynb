```python
pdf_path = "/content/drive/My Drive/Project 7/Report-11-02-012017.pdf"
```

```python
income_tables = tabula.read_pdf(
    pdf_path,
    pages = "20",
    multiple_tables = True,
    lattice= True
)

for i, table in enumerate(income_tables):
  print(f"Table {i}:")
  print(table.head())
  print("---")
```

```python
crop_tables = tabula.read_pdf(
    pdf_path,
    pages = "53",
    multiple_tables = True,
    stream= True,
    guess= False,
    area= [50, 0, 1000, 1200]
)

for i, table in enumerate(crop_tables):
 print(f"Table {i}:")
 print(table.head())
 print("---")
```
